<h2>2. K-1 OneHotEncoding</h2>

```python
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True)

# What it does
#     Performs One-Hot Encoding
#     Drops one dummy column per categorical feature
#     Keeps k−1 columns instead of k
```

<h2>OneHotEncoding with Top Categories</h2>


```python
counts = df['brand'].value_counts()
print(counts)
```

```python
df['brand'].nunique()
threshold = 100
```

```python
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl, 'uncommon')).sample(5)
```
