<h2>use_cols parameter</h2>

usecols tells Pandas which columns to read from the file
and which columns to ignore.

    Save memory
    Speed up large file loading

<h2>Skiprows</h2>

skiprows tells Pandas which rows to ignore while reading a file.

pd.read_csv('data.csv', skiprows=2)
✔ Skips first 2 rows
✔ Reading starts from row index 2

pd.read_csv('data.csv', skiprows=[0, 2, 5])
✔ Skips rows with indices 0, 2, and 5

pd.read_csv('data.csv', skiprows=lambda x: x < 3)
✔ Skips first 3 rows dynamically


```python
pd.read_csv('aug_train.csv',skiprows=[2,3,5])
```

<h2>Encoding Parameter</h2>

encoding tells Pandas how text is stored in the file (character format). 

If Pandas uses the wrong encoding:
❌ File won’t load
❌ Text becomes corrupted
❌ ML text features break

**❌ Error you may see**:-UnicodeDecodeError: 'utf-8' codec can't decode byte
**✅ Fix**:-df = pd.read_csv('data.csv', encoding='latin1')



<h2>bad line” error?</h2>

A bad line is a row in a CSV file that does not follow the expected column structure

*ParserError: Error tokenizing data. C error: Expected 5 fields in line 100, saw 7* -->👉 Pandas expected 5 columns, but found 7 in that row.

<h3>4️⃣ Correct Ways to Handle Bad Lines ✅</h3>

Skip bad lines:-
    pd.read_csv('data.csv', on_bad_lines='skip')

Warn and skip
    pd.read_csv('data.csv', on_bad_lines='warn')


<h3>❌ Deprecated (Old Pandas)</h3>
    error_bad_lines=False
    warn_bad_lines=True

<h2>Handling dates</h2>

by default we pass dates as string but if we pass it as string it will loss functionality of date

*parse_dates tells Pandas to convert date/time columns into datetime objects while reading the file.*

<h3>✅ Case 1: Parse a single date column</h3>

df = pd.read_csv(
    'sales.csv',
    parse_dates=['order_date']
)

<h3>✅ Case 2: Parse multiple date columns</h3>

df = pd.read_csv(
    'data.csv',
    parse_dates=['start_date', 'end_date']
)

<h2>na_values</h2>

na_values tells Pandas which values in your CSV should be considered as NaN (missing).

pd.read_csv('file.csv', na_values=...)

👉 Instead of only using empty cells as missing, you can define custom missing values.

<h3>✅ Case 1: Single value</h3>
    df = pd.read_csv('data.csv', na_values='?')-->✔ Every ? becomes NaN

<h3>✅ Case 2: Multiple values (list)</h3>
    df = pd.read_csv('data.csv', na_values=['?', 'NA', 'NULL', '-'])
    ✔ All of these are treated as missing

<h3>✅ Case 3: Column-specific missing values (dict)</h3>

na_dict = {
    'age': ['NA', '-1'],
    'city': ['Unknown', '']
}

df = pd.read_csv('data.csv', na_values=na_dict)

✔ Custom missing values per column
