```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/StudentsPerformance.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

Visualising the relationship b/w data

```python
avg_scores=df.groupby('gender')[['math score','reading score','writing score']].mean() #meaning we are grouping by gender (male,female)

#for each gender group , keep only these three score columns

avg_scores.plot(kind='bar',figsize=(10,5))

plt.title('Avg Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Avg Score')
plt.show()
```

we can see that males are performing better in maths but females are performing better in reading score, and writing score

```python
#boxplot

#math score
import seaborn as sns

sns.boxplot(x='gender',y='math score',data=df)

plt.title('Math Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Math Score')
plt.show()
```

```python
sns.boxplot(x='gender',y='reading score',data=df)

plt.title('Reading Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Reading Score')
plt.show()
```

```python
#for writing score

sns.boxplot(x='gender',y='writing score',data=df)

plt.title('Writing Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Writing Score')
plt.show()
```

```python
avg_scores=df.groupby('race/ethnicity')[['math score','reading score','writing score']].mean() #meaning we are grouping by gender (male,female)

#for each gender group , keep only these three score columns
plt.figure(figsize=(6,4))
avg_scores.plot(kind='bar',figsize=(10,5))

plt.title('Avg Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Avg Score')
plt.show()
```

as group is increasing avg performance is increasing

now visuailising relationship b/w scores itself

```python
scores=df[['math score','reading score','writing score']]

scores.corr()
```

```python
#visualising through heatmaps

corr_matrix=scores.corr()

sns.heatmap(corr_matrix,annot=True,cmap='coolwarm')
plt.title('Correlation B/w scores')
plt.show()
```

so it means if he is scoring in higher in one subject then he will also score higher in another subject also
