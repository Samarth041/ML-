```python
import mysql.connector

# Connect to MySQL (without specifying a database)
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Raj@123456"
)

cursor = conn.cursor()

# Make sure your target database exists
cursor.execute("CREATE DATABASE IF NOT EXISTS sqlpractice_db;")
cursor.execute("USE sqlpractice_db;")

# Read the SQL file
with open("sqlpractice.sql", "r") as f:
    sql_commands = f.read()

# Remove any USE statements to avoid unknown database errors
lines = sql_commands.splitlines()
clean_commands = []
for line in lines:
    line = line.strip()
    if line.upper().startswith("USE "):
        continue  # skip USE statements
    clean_commands.append(line)

# Join back into commands and split by semicolon
sql_commands_clean = "\n".join(clean_commands).split(';')

# Execute each command
for command in sql_commands_clean:
    command = command.strip()
    if command:  # skip empty commands
        try:
            cursor.execute(command)
        except Exception as e:
            print(f"Skipping command due to error: {e}")

conn.commit()
print("SQL file executed successfully!")
```
