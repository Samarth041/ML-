<h2>Working with dates</h2>

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

<h4>1. Extract year</h4>

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

<h4>2. Extract Month</h4>

```python
date['date_month_no'] = date['date'].dt.month

date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

<h4>Extract Days</h4>

```python
date['date_day'] = date['date'].dt.day #mahine ka kon sa din chl raha hai

date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek #hafte ka kon sa din chal raha hai

date.head()
```

```python

# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

<h4>Extract week of the year</h4>

🔹 Why isocalendar()?

ISO standard defines:

    Weeks start on Monday
    Week 1 = first week with at least 4 days in the new year

This is industry-standard.

```python
date['date_week'] = date['date'].dt.isocalendar().week.astype(int) #kon sa week hai


date.drop(columns=['product_id','city_id','orders']).head()
```

<h2>Extract Quarter</h2>

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

<h2>Extract Semester</h2>

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

<h2>Extract Time elapsed between dates
</h2>

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed

months_passed = np.round(
    (today - date['date']) / np.timedelta64(1, 'D') / 30,
    0
)

months_passed
```
