```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('StudentsPerformance.csv')
```

Visualising the relationship b/w data

```python
avg_scores=df.groupby('gender')[['math score','reading score','writing score']].mean() #meaning we are grouping by gender (male,female)

#for each gender group , keep only these three score columns

avg_scores.plot(kind='bar',figsize=(10,5))

plt.title('Avg Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Avg Score')
plt.show()
```

we can see that males are performing better in maths but females are performing better in reading score, and writing score

```python
#boxplot

#math score
import seaborn as sns

sns.boxplot(x='gender',y='math score',data=df)

plt.title('Math Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Math Score')
plt.show()
```

```python
sns.boxplot(x='gender',y='reading score',data=df)

plt.title('Reading Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Reading Score')
plt.show()
```

```python
scores=df[['math score','reading score','writing score']]

scores.corr()
```

```python
#visualising through heatmaps

corr_matrix=scores.corr()

sns.heatmap(corr_matrix,annot=True,cmap='coolwarm')
plt.title('Correlation B/w scores')
plt.show()
```

```python
low_scorers=(df['math score']<40).sum()

toppers=(df['math score']>=90).sum()

total_students=len(df)

print("Low scorers ", low_scorers)

print("Toppers ",toppers)

print("low scorers % ",low_scorers/total_students*100)
print("Toppers % ", toppers/total_students*100 )
```

```python
sns.boxplot(x=df['math score'])
plt.show()
```

```python
#defining categorical columns

cat_cols=['gender',
          'race/ethnicity',
          'parental level of education',
          'lunch',
          'test preparation course'
         ]
```

```python
#avg math scores for each category

for col in cat_cols:
    print("\n=========================")
    print("Feature",col)
    print("=====================")
    print(df.groupby(col)['math score'].mean().sort_values(ascending=False))
```

```python
for col in cat_cols:
    grouped = df.groupby(col)['math score'].mean()

    impact = grouped.max() - grouped.min()

    print("\nFeature:", col)
    print("Impact (max - min):", round(impact, 2))
    print(grouped.sort_values(ascending=False))
```

```python
#visully

for col in cat_cols:
    plt.figure(figsize=(8,4))
    sns.barplot(x=col,y='math score',data=df,estimator='mean')
    plt.title(f"Avergae Math Score by {col}")
    plt.xticks(rotation=45)
    plt.show()
```
